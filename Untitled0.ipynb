```python
import tensorflow as tf

# Drive'ındaki klasörün tam yolu
veri_yolu = '/content/drive/MyDrive/TrashType_Image_Dataset'

# Eğitim veri setini içeri alıyoruz
train_ds = tf.keras.utils.image_dataset_from_directory(
    veri_yolu,
    validation_split=0.2,       # %20'sini test/doğrulama için ayırır
    subset="training",
    seed=123,
    image_size=(224, 224),       # Resimleri modelin okuyacağı standart boyuta getirir
    batch_size=32                # Her adımda 32'şer resim işler
)
```

```python
# Doğrulama (Validation) veri setini içeri alıyoruz
val_ds = tf.keras.utils.image_dataset_from_directory(
    veri_yolu,
    validation_split=0.2,
    subset="validation",    # Bu sefer subset kısmını 'validation' yapıyoruz
    seed=123,
    image_size=(224, 224),
    batch_size=32
)
```

```python
from tensorflow.keras import layers, models

# 6 farklı atık türü (sınıfı) olduğu için
num_classes = 6

# Basit bir CNN mimarisi oluşturuyoruz
model = models.Sequential([
  # Resim piksel değerlerini 0-255 arasından 0-1 arasına normalize eder
  layers.Rescaling(1./255, input_shape=(224, 224, 3)),

  # Özellik çıkarma katmanları (Resimdeki çizgileri, şekilleri yakalar)
  layers.Conv2D(16, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Conv2D(32, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Conv2D(64, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),

  # Sınıflandırma katmanları (Öğrenilen özellikleri 6 sınıfa bağlar)
  layers.Flatten(),
  layers.Dense(128, activation='relu'),
  layers.Dense(num_classes)
])

# Modeli derleme (Öğrenme stratejisini belirliyoruz)
model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

# Modelin özet yapısını ekrana yazdıralım
model.summary()
```

```python
# Modelin veriyi baştan sona kaç tur döneceğini belirliyoruz
# İlk deneme için 10 tur (epochs) idealdir, duruma göre sonra artırabilirsin
epochs = 10

# Eğitimi başlatıyoruz
history = model.fit(
  train_ds,
  validation_data=val_ds,
  epochs=epochs
)
```

```python
import matplotlib.pyplot as plt

# Başarı oranlarını grafiğe döküyoruz
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Eğitim Başarısı')
plt.plot(history.history['val_accuracy'], label='Doğrulama Başarısı')
plt.xlabel('Tur (Epoch)')
plt.ylabel('Başarı Oranı')
plt.legend()
plt.title('Eğitim ve Doğrulama Başarısı')

# Kayıp (Loss) değerlerini grafiğe döküyoruz
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Eğitim Kaybı')
plt.plot(history.history['val_loss'], label='Doğrulama Kaybı')
plt.xlabel('Tur (Epoch)')
plt.ylabel('Kayıp (Loss)')
plt.legend()
plt.title('Eğitim ve Doğrulama Kaybı')

plt.show()
```

```python
from tensorflow.keras import layers, models

# Resimleri rastgele sağa/sola çevirip döndürerek veri çeşitliliğini artırıyoruz
data_augmentation = tf.keras.Sequential([
  layers.RandomFlip("horizontal_and_vertical"),
  layers.RandomRotation(0.2),
])

num_classes = 6

# Geliştirilmiş yeni model mimarimiz
model_v2 = models.Sequential([
  data_augmentation, # Veri artırma katmanı en başta
  layers.Rescaling(1./255, input_shape=(224, 224, 3)),

  layers.Conv2D(16, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),

  layers.Conv2D(32, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),

  layers.Conv2D(64, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),

  # Ezberi engellemek için Dropout ekliyoruz (%20 nöron kapatma)
  layers.Dropout(0.2),

  layers.Flatten(),
  layers.Dense(128, activation='relu'),

  # Bir dropout da buraya ekliyoruz
  layers.Dropout(0.2),
  layers.Dense(num_classes)
])

# Yeni modeli derleme
model_v2.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=['accuracy']
)

# Yeni modeli eğitme (Yine 10 tur deneyelim)
history_v2 = model_v2.fit(
  train_ds,
  validation_data=val_ds,
  epochs=10
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from tensorflow.keras.preprocessing import image

# 1. Klasör isimlerini (Plastik, Cam, Metal vb.) otomatik olarak çekiyoruz
class_names = train_ds.class_names
print("Modelin bildiği sınıflar:", class_names)

# 2. Resim yükleme ve tahmin fonksiyonu
def resim_test_et(resim_yolu):
    # Resmi modelin okuyabileceği 224x224 boyutuna getiriyoruz
    img = image.load_img(resim_yolu, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array = tf.expand_dims(img_array, 0) # Yapay zekanın beklediği matris boyutuna çevirir

    # Model v2 kullanarak tahmin yapıyoruz
    predictions = model_v2.predict(img_array)
    score = tf.nn.softmax(predictions[0])

    # Sonucu ekrana yazdırıyoruz
    tahmin_edilen_sinif = class_names[np.argmax(score)]
    guven_orani = 100 * np.max(score)

    print("\n--------------------------------------------------")
    print(f"🤖 YAPAY ZEKA TAHMİNİ: {tahmin_edilen_sinif.upper()}")
    print(f"🎯 GÜVEN ORANI: %{guven_orani:.2f}")
    print("--------------------------------------------------\n")

    # Resmi ekranda gösteriyoruz
    plt.imshow(img)
    plt.axis('off')
    plt.show()

# 3. Bilgisayarından resim seçmenizi sağlayan buton
print("\nLütfen test etmek istediğiniz bir atık resmi seçin (jpg veya png):")
yuklenen_dosya = files.upload()

# Yüklediğin resmi fonksiyona gönderiyoruz
for dosya_adi in yuklenen_dosya.keys():
    resim_test_et(dosya_adi)
```
